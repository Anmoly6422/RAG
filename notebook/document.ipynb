```python
sample_texts = {
    "../data/text_files/python_intro.txt": """Python Introduction

Python is a high-level, interpreted programming language known for its simple and readable syntax. It is widely used in web development, data science, artificial intelligence, machine learning, automation, and scripting.

Python uses indentation to define blocks of code instead of curly braces. Variables do not require explicit type declarations.

For example:

name = "Anmol"
age = 21

print("Hello", name)
print("Age:", age)

Python provides built-in data structures such as lists, tuples, sets, and dictionaries.

Example of a list:

languages = ["Python", "JavaScript", "C++"]

for language in languages:
    print(language)

Functions allow developers to organize reusable code:

def greet(name):
    return f"Hello, {name}!"

print(greet("Anmol"))

Python has a large ecosystem of libraries and frameworks. It is commonly used for web development, data science, machine learning, automation, and AI applications.

Python is also useful for building Retrieval-Augmented Generation (RAG) applications. In a RAG system, Python can be used to load documents, split them into chunks, create embeddings, store vectors, retrieve relevant information, and generate answers using a language model.

Learning Python fundamentals such as variables, data types, conditions, loops, functions, modules, and classes provides a strong foundation for building RAG applications.
""",

    "../data/text_files/machine_learning.txt": """Introduction to Machine Learning

Machine Learning (ML) is a branch of artificial intelligence that allows computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

Traditional programming generally uses rules and data to produce an output. Machine learning uses data and, in many cases, expected outputs to learn a model that can make predictions on new data.

Types of Machine Learning

There are three major types of machine learning:

1. Supervised Learning

Supervised learning uses labeled data, where the input and expected output are known. The model learns the relationship between inputs and outputs.

Common supervised learning tasks include:

- Classification: Predicting a category or class.
- Regression: Predicting a numerical value.

Examples include email spam detection, house price prediction, and disease classification.

2. Unsupervised Learning

Unsupervised learning uses data without predefined labels. The model tries to discover patterns, structures, or groups within the data.

A common unsupervised learning technique is clustering. Clustering groups similar data points together.

Examples include customer segmentation, document grouping, and identifying unusual patterns.

3. Reinforcement Learning

Reinforcement learning is a method in which an agent learns by interacting with an environment. The agent receives rewards or penalties based on its actions and tries to maximize its total reward over time.

Reinforcement learning is used in robotics, game-playing systems, and decision-making applications.

Common Machine Learning Algorithms

Some popular machine learning algorithms include:

- Linear Regression: Used to predict continuous numerical values.
- Logistic Regression: Commonly used for classification.
- Decision Trees: Make predictions using a tree-like structure of decisions.
- Random Forest: Combines multiple decision trees to improve predictions.
- K-Nearest Neighbors (KNN): Predicts outcomes based on nearby data points.
- Support Vector Machines (SVM): Finds boundaries between different classes.
- K-Means Clustering: Groups data points into clusters.

Machine Learning Workflow

A typical machine learning project follows several steps:

1. Collect data from relevant sources.
2. Clean the data by handling missing values and incorrect entries.
3. Explore the data to understand its patterns and relationships.
4. Select useful features for training the model.
5. Split the dataset into training and testing sets.
6. Train the machine learning model.
7. Evaluate the model using suitable metrics.
8. Improve and deploy the model when it performs sufficiently well.

Training and Testing

The training dataset is used by the model to learn patterns. The testing dataset is used to evaluate how well the model performs on data it has not seen during training.

A model that performs well on training data but poorly on new data may be overfitting. Overfitting occurs when a model learns the training data too closely, including noise, instead of learning general patterns.

Underfitting occurs when a model is too simple to capture important patterns in the data.

Machine Learning and Python

Python is widely used in machine learning because it has many useful libraries and frameworks.

Popular machine learning libraries include:

- NumPy: Used for numerical operations and arrays.
- pandas: Used for data manipulation and analysis.
- Matplotlib: Used for data visualization.
- Scikit-learn: Provides traditional machine learning algorithms.
- TensorFlow: Used for building and training machine learning models.
- PyTorch: Used for deep learning and scientific computing.

Machine Learning and RAG

Machine learning concepts are related to Retrieval-Augmented Generation (RAG) applications. RAG systems commonly use embedding models to convert text into numerical vectors. These vectors represent the semantic meaning of text and can be used to find relevant information.

In a RAG pipeline, documents are loaded, divided into smaller chunks, converted into embeddings, and stored in a vector database. When a user asks a question, the system retrieves relevant chunks and provides them to a language model to generate a grounded response.

Machine learning provides the foundation for many technologies used in modern artificial intelligence, including recommendation systems, natural language processing, computer vision, and RAG applications.
"""
}

for filepath, content in sample_texts.items():
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(content)

print("Sample text files created")
```
